```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

x = np.array([
    2.1, 2.3, 2.5, 3.0, 3.2,
    3.9, 4.0, 4.2, 4.7, 5.0, 3.9
])

y = np.array([
    49000, 56000, 47700, np.nan, 49800,
    67700, 70300, 65500, 68200, 73400, 67700
])

df = pd.DataFrame({
    "Experience": x,
    "Salary": y
})

print(df)
```

```python
plt.plot(x, y, marker='D')

plt.xlabel("Experience")
plt.ylabel("Salary")

plt.title("Original Data - Experience Vs Salary")

plt.grid()
plt.show()
```

```python
print(df.mean())
print(df.median())
```

```python
print(df.describe())
```

```python
print(df.isnull())
print(df.isnull().sum())
```

```python
print(df.duplicated())
print(df.duplicated().sum())
```

```python
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

print(df)
```

```python
df = df.drop_duplicates()

print(df)

print(df.isnull().sum())
print(df.duplicated().sum())
```

```python
x = df[["Experience"]]
y = df["Salary"]

plt.plot(x, y, marker='D')

plt.xlabel("Experience")
plt.ylabel("Salary")

plt.title("Original Data - Experience Vs Salary")

plt.grid()
plt.show()
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)

print(x_test.shape)
print(x_train.shape)
print(y_test.shape)
print(y_train.shape)
```

```python
model = LinearRegression()

model.fit(x_train, y_train)

print(model)
```

```python
y_pred = model.predict(x_test)

print("Prediction value:", y_pred)
```

```python
r2 = r2_score(y_test, y_pred)

print("R2 Score:", r2)
```

```python
experience = np.array([[3]])

salary = model.predict(experience)

print("Predicted Salary:", salary[0])
```
